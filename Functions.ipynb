# 1. What is the difference between a function and a method in Python?
  - In Python:

A function is a block of reusable code that performs a specific task. It is independent and can be called by its name.

A method is similar to a function, but it is associated with an object (usually a class instance) and called on that object. A method always takes at least one argument, usually self, referring to the object itself.

In short:

Function: standalone code block.

Method: function that belongs to an object (like a class instance).

Example.

```python
# Function
def greet(name):
    return f"Hello, {name}!"

print(greet("Alice"))

# Method (inside a class)
class Person:
    def greet(self):
        return f"Hello, {self.name}!"

    def __init__(self, name):
        self.name = name

p = Person("Bob")
print(p.greet())
```

# 6. Explain the concept of generators in Python and how they are defined.

  - Generators in Python

A generator in Python is a special type of iterator that allows you to iterate over a sequence of values without storing the entire sequence in memory. Instead of returning all values at once, a generator yields one value at a time using the yield keyword.

Generators are useful for lazy evaluation, meaning values are generated only when needed. This makes them memory-efficient, especially when working with large datasets or infinite sequences.


-How Generators Are Defined

Generators can be defined in two main ways:

Using a generator function (with yield)

Using a generator expression (similar to list comprehensions)

Example.

```python
def count_up_to(n):
    count = 1
    while count <= n:
        yield count
        count += 1
for num in count_up_to(5):
    print(num)
```

# 7. What are the advantages of using generators over regular functions?

 - Advantages of Using Generators Over Regular Functions
Memory Efficiency:
Generators yield one item at a time and do not store the entire sequence in memory, unlike regular functions that return complete data structures (e.g., lists).

Lazy Evaluation:

Values are produced only when needed, making generators ideal for handling large data sets or infinite sequences.

Improved Performance:

Since data is generated on the fly, performance is often better for I/O-bound or memory-heavy operations.

Clean Syntax and Easy to Implement:

Writing a generator using yield is more readable and concise than manually implementing an iterator class.

- Example

```python
#regular function
def get_squares_list(n):
    result = []
    for i in range(n):
        result.append(i * i)
    return result

print(get_squares_list(5))
```

```python
#generator function
def generate_squares(n):
    for i in range(n):
        yield i * i

for square in generate_squares(5):
    print(square)
```

# 10. What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?

 - Difference Between map(), reduce(), and filter() in Python

These three functions are part of functional programming tools in Python and are used to process iterables. Here's how they differ:

 - map()

Purpose: Applies a function to each item in the iterable

Returns: A new iterable with transformed items

Function Signature:map(function, iterable)



 - filter()

Purpose: Filters items based on a condition (function returns True or False)

Returns: A filtered iterable with only the items that satisfy the condition

Function Signature:filter(function, iterable)



 - reduce()

Purpose: Reduces the iterable to a single cumulative value by repeatedly applying a function

Returns: A single value

Function Signature:reduce(function, iterable)


```python
# Map
numbers = [1, 2, 3, 4]
squared = list(map(lambda x: x ** 2, numbers))
print(squared)  # Output: [1, 4, 9, 16]
```

```python
# filter
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
print(even_numbers)  # Output: [2, 4]
```

```python
# reduce
from functools import reduce
product = reduce(lambda x, y: x * y, numbers)
print(product)  # Output: 24
```

# 1. Write a Python function that takes a list of numbers as input and returns the sum of all even numbers in the list.

 - Python function that takes a list of numbers and returns the sum of all even numbers:

 example.


```python
def sum_of_even_numbers(numbers):
    even_sum = 0
    for num in numbers:
        if num % 2 == 0:
            even_sum += num
    return even_sum

```

# 4. Write a Python function that checks if a given number is prime or not from 1 to 200.

 - Here's a Python function that checks if a given number is prime or not, specifically for numbers between 1 and 200:




```python
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

# Example usage: Check if numbers from 1 to 200 are prime
for num in range(1, 201):
    if is_prime(num):
        print(f"{num} is prime")
```

# 5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms.

 - Here's a Python iterator class that generates the Fibonacci sequence up to a specified number of terms:


```python
class FibonacciIterator:
    def __init__(self, n):
        self.n = n  # Number of terms to generate
        self.a, self.b = 0, 1  # First two terms of Fibonacci
        self.count = 0  # Count to track number of terms generated

    def __iter__(self):
        return self  # Return the iterator object itself

    def __next__(self):
        if self.count >= self.n:
            raise StopIteration  # Stop iteration once we reach the specified number of terms
        # Generate the next Fibonacci term
        fib = self.a
        self.a, self.b = self.b, self.a + self.b
        self.count += 1
        return fib

# Example usage:
n_terms = 10
fibonacci_sequence = FibonacciIterator(n_terms)

# Print Fibonacci sequence up to 10 terms
for num in fibonacci_sequence:
    print(num)
```

# 6. Write a generator function in Python that yields the powers of 2 up to a given exponent.
  
- Here’s a Python generator function that yields the powers of 2 up to a given exponent:




```python
def powers_of_two(exponent):
    for i in range(exponent + 1):
        yield 2 ** i

# Example usage: Get powers of 2 up to 5 (2^0, 2^1, ..., 2^5)
for power in powers_of_two(5):
    print(power)
```

# 7. Implement a generator function that reads a file line by line and yields each line as a string.

  
  - Here’s a Python generator function that reads a file line by line and yields each line as a string:




```python
def read_file_line_by_line(file_path):
    with open(file_path, 'r') as file:
        for line in file:
            yield line.strip()  # Using strip to remove any trailing newline characters

# Example usage:
file_path = 'sample.txt'  # Replace with the path to your file
for line in read_file_line_by_line(file_path):
    print(line)
```

# 9. Write a Python program that uses `map()` to convert a list of temperatures from Celsius to Fahrenheit.

 - Here’s a Python program that uses the map() function to convert a list of temperatures from Celsius to Fahrenheit:



```python
# List of temperatures in Celsius
celsius_temps = [0, 20, 37, 100, -10]

# Function to convert Celsius to Fahrenheit
def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32

# Use map() to apply the conversion to each temperature in the list
fahrenheit_temps = list(map(celsius_to_fahrenheit, celsius_temps))

# Print the result
print(fahrenheit_temps)
```

# 10. Create a Python program that uses `filter()` to remove all the vowels from a given string.

 - Here’s a Python program that uses the filter() function to remove all vowels from a given string:



```python
# Function to check if a character is a vowel
def is_not_vowel(char):
    return char.lower() not in 'aeiou'

# Given string
input_string = "Hello World"

# Use filter() to remove vowels
filtered_string = ''.join(filter(is_not_vowel, input_string))

# Print the result
print(filtered_string)
```
